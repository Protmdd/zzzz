```python
constantes = [c for c in atributos if datos[c].nunique() <= 1]

atributos = [c for c in atributos if c not in constantes]

print("Variables constantes eliminadas:", constantes)
print("Atributos que quedan:", atributos)
```

<font color=blue>3.5 Valores atípicos</font>


```python
X = datos[atributos].copy()

q1 = X.quantile(0.25)
q3 = X.quantile(0.75)
iqr = q3 - q1

atipicos = ((X < q1 - 1.5 * iqr) | (X > q3 + 1.5 * iqr)).sum()

display(pd.DataFrame({
    "n_atipicos": atipicos,
    "porcentaje": (100 * atipicos / len(X)).round(1)
}))

ax = sns.boxplot(data=(X - X.mean()) / X.std(), orient="h")
ax.set(title="Boxplot de los atributos estandarizados", xlabel="Valor z")
plt.show()
```

<font color=green>18. Probamos eps creciente</font>


```python
comparacion_eps = []

for eps in [1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0]:   # COMPLETAR: valores alrededor del codo del gráfico anterior

    modelo = DBSCAN(eps=eps, min_samples=min_samples)

    labels = modelo.fit_predict(X_std)

    etiquetas = set(labels)
    n_clusters = len(etiquetas) - (1 if -1 in etiquetas else 0)

    porcentaje_ruido = np.mean(labels == -1) * 100

    sin_ruido = labels != -1

    if n_clusters >= 2:
        sil = silhouette_score(X_std[sin_ruido], labels[sin_ruido])
        db = davies_bouldin_score(X_std[sin_ruido], labels[sin_ruido])
    else:
        sil = np.nan
        db = np.nan

    comparacion_eps.append({
        "eps": eps,
        "n_clusters": n_clusters,
        "ruido_%": porcentaje_ruido,
        "silhouette_sin_ruido": sil,
        "davies_bouldin_sin_ruido": db
    })

comparacion_eps = pd.DataFrame(comparacion_eps)

display(comparacion_eps.round(3))
```

```python
ax = sns.lineplot(data=comparacion_eps, x="eps", y="n_clusters", marker="o")
ax.set(title="DBSCAN: clusters encontrados según eps", xlabel="eps", ylabel="Número de clusters")
plt.show()

ax = sns.lineplot(data=comparacion_eps, x="eps", y="ruido_%", marker="o")
ax.set(title="DBSCAN: porcentaje de ruido según eps", xlabel="eps", ylabel="% de ruido")
plt.show()
```

<font color=green>19. Combinaciones de `min_samples` y `eps`</font>

```python
combinaciones = []

for ms in [3, 5, 8, 10, 15, 20]:
    for eps in [0.75, 1.0, 1.25, 1.5, 2.0, 3.0]:   # COMPLETAR

        labels = DBSCAN(eps=eps, min_samples=ms).fit_predict(X_std)

        etiquetas = set(labels)
        n_clusters = len(etiquetas) - (1 if -1 in etiquetas else 0)

        combinaciones.append({
            "min_samples": ms,
            "eps": eps,
            "n_clusters": n_clusters,
            "ruido_%": np.mean(labels == -1) * 100
        })

combinaciones = pd.DataFrame(combinaciones)

tabla_clusters = combinaciones.pivot(index="min_samples", columns="eps", values="n_clusters")
tabla_ruido = combinaciones.pivot(index="min_samples", columns="eps", values="ruido_%")

sns.heatmap(tabla_clusters, annot=True, fmt=".0f", cmap="Blues")
plt.title("Número de clusters según min_samples y eps")
plt.show()

sns.heatmap(tabla_ruido, annot=True, fmt=".0f", cmap="Oranges")
plt.title("Porcentaje de ruido según min_samples y eps")
plt.show()
```

<font color=blue>PARTE VI. Observaciones nuevas</font>

Los modelos ya están ajustados. A una observación nueva **no** se le vuelve a ajustar nada:

- escalador: `transform` (nunca `fit_transform`);
- K-means: `predict`, que le asigna el centroide más cercano;
- DBSCAN: **no tiene** `predict`.
<font color=green>22. Una observación nueva</font>

```python
nuevo = pd.DataFrame({
    "col1": [0],   # COMPLETAR: un valor por cada atributo, en el mismo orden
    "col2": [0],
    "col3": [0],
})

# Si el enunciado define una variable calculada, crearla aquí. Ejemplo:
# nuevo["ratio"] = 100 * (nuevo["deuda"]) / nuevo["ingreso"]

# Aplicar la misma transformación que a X. Ejemplo: nuevo["ingreso"] = np.log1p(nuevo["ingreso"])

display(nuevo)
```

```python
x_nuevo = scaler_std.transform(nuevo[atributos])

seg = kmeans.predict(x_nuevo)[0]

print("Segmento asignado:", seg)
print("Distancia a cada centroide:", kmeans.transform(x_nuevo).round(2))
print("Tasa de target del segmento:", perfil.loc[seg, "tasa_target"])
```

<font color=green>23. ¿Qué tan lejos queda de su segmento?</font>

`predict` siempre asigna el centroide más cercano, aunque la observación sea muy distinta de todas. Para detectarlas comparamos su distancia al centroide con la de la base original.

```python
dist_base = kmeans.transform(X_std).min(axis=1)

# El 95 % de la base está más cerca que este valor de su centroide.
umbral = np.percentile(dist_base, 95)

dist_nuevo = kmeans.transform(x_nuevo).min()

print("Distancia a su centroide:", round(dist_nuevo, 2), "| umbral:", round(umbral, 2))
print("¿Revisión manual?", dist_nuevo > umbral)
```

<font color=green>24. Varias observaciones nuevas</font>

```python
nuevos = pd.read_csv("NUEVOS.csv")   # COMPLETAR: archivo con las observaciones nuevas

# Aplicar la misma transformación que a X

X_nuevos_std = scaler_std.transform(nuevos[atributos])

nuevos["segmento"] = kmeans.predict(X_nuevos_std)
nuevos["dist_centroide"] = kmeans.transform(X_nuevos_std).min(axis=1)
nuevos["revisar"] = nuevos["dist_centroide"] > umbral

display(nuevos["segmento"].value_counts().sort_index().to_frame("n_nuevos"))
print("Para revisión manual:", nuevos["revisar"].sum())
```

<font color=green>25. DBSCAN no tiene `predict`</font>

Regla sencilla: si la observación nueva está a distancia menor o igual que `eps` de algún **punto núcleo**, toma el cluster de ese núcleo; si no, es ruido.

```python
nucleos = X_std[dbscan.core_sample_indices_]
cluster_nucleos = dbscan.labels_[dbscan.core_sample_indices_]

vecino = NearestNeighbors(n_neighbors=1)
vecino.fit(nucleos)

dist_nucleo, idx_nucleo = vecino.kneighbors(X_nuevos_std)

nuevos["dbscan"] = np.where(
    dist_nucleo[:, 0] <= dbscan.eps,
    cluster_nucleos[idx_nucleo[:, 0]],
    -1
)

display(nuevos["dbscan"].value_counts().sort_index().to_frame("n_nuevos"))
```
